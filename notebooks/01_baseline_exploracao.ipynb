# Estudo e Avaliação de Técnicas de Redução do Número de Iterações do K-Means
## Notebook 01: Exploração da Linha de Base (Baseline Lloyd) e Métricas

Este notebook demonstra o funcionamento da linha de base do K-Means com distância euclidiana quadrática e o comparativo inicial com técnicas de redução de iterações.

```python
import sys
import os

# Adiciona o diretório src ao path
sys.path.insert(0, os.path.abspath("../src"))

from kmeans_reduction import StandardKMeans, ModifiedKMeans, FastCodebookKMeans
from kmeans_reduction.applications.benchmark_clustering import generate_synthetic_blobs
from kmeans_reduction.metrics import compare_kmeans_results

print("Módulos importados com sucesso!")
```

### 1. Geração de Dados Sintéticos com Agrupamentos Conhecidos

```python
X, y_true = generate_synthetic_blobs(
    n_samples=300,
    n_features=2,
    centers=3,
    cluster_std=1.0,
    random_state=42
)
print(f"Total de amostras geradas: {len(X)}")
print(f"Exemplo de ponto: {X[0]}")
```

### 2. Ajuste do K-Means Padrão (Linha de Base)

```python
baseline = StandardKMeans(n_clusters=3, max_iter=100, random_state=42)
baseline.fit(X)

print(f"Iterações executadas: {baseline.n_iter_}")
print(f"Inércia final: {baseline.inertia_:.4f}")
print(f"Tempo de execução: {baseline.result_.elapsed_time_sec * 1000:.2f} ms")
print(f"Centróides encontrados:\n{baseline.cluster_centers_}")
```

### 3. Comparação com Técnicas de Redução de Iterações

```python
mod_kmeans = ModifiedKMeans(n_clusters=3, activity_threshold=0.01, random_state=42).fit(X)
fast_kmeans = FastCodebookKMeans(n_clusters=3, init_strategy="pattern_spread", random_state=42).fit(X)

print("--- Comparativo contra Baseline ---")
comp_mod = compare_kmeans_results(baseline.result_, mod_kmeans.result_)
print(f"Modified K-Means -> Iterações: {comp_mod['candidate_iterations']} (Redução: {comp_mod['iteration_reduction_pct']:.1f}%)")

comp_fast = compare_kmeans_results(baseline.result_, fast_kmeans.result_)
print(f"Fast Codebook   -> Iterações: {comp_fast['candidate_iterations']} (Redução: {comp_fast['iteration_reduction_pct']:.1f}%)")
```
